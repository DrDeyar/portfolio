## Homework Hack — High School Daily Routine Dashboard


```python
# CODE_RUNNER: Homework Hack - Build a high school daily routine dashboard

student = {
    "logged_in": True,
    "homework_finished": False,
    "lunch_packed": True,
    "practice_today": True
}

def daily_routine(person):
    # The account check comes first, before any schedule decisions.
    if person["logged_in"]:
        if person["homework_finished"]:
            if person["practice_today"]:
                if person["lunch_packed"]:
                    return "Homework is done. Eat the packed lunch, then head to practice."
                else:
                    return "Homework is done. Get lunch before heading to practice."
            else:
                if person["lunch_packed"]:
                    return "Homework is done. Eat the packed lunch and use the extra time."
                else:
                    return "Homework is done. Get lunch and enjoy the free afternoon."
        else:
            if person["practice_today"]:
                return "Finish the most important homework first, then get ready for practice."
            else:
                return "No practice today, so finish the homework before free time."
    else:
        return "Log in first before checking the rest of the daily plan."

print(daily_routine(student))

# Test values that reach the meaningful outcomes.
tests = [
    {"logged_in": False, "homework_finished": False, "lunch_packed": False, "practice_today": False},
    {"logged_in": True, "homework_finished": True, "lunch_packed": True, "practice_today": True},
    {"logged_in": True, "homework_finished": True, "lunch_packed": False, "practice_today": True},
    {"logged_in": True, "homework_finished": True, "lunch_packed": True, "practice_today": False},
    {"logged_in": True, "homework_finished": True, "lunch_packed": False, "practice_today": False},
    {"logged_in": True, "homework_finished": False, "lunch_packed": True, "practice_today": True},
    {"logged_in": True, "homework_finished": False, "lunch_packed": True, "practice_today": False},
]

for number, test in enumerate(tests, start=1):
    print(f"Test {number}: {daily_routine(test)}")
```

### Why the checks are nested

The schedule checks depend on the student being logged in, so there is no reason to check homework, lunch, or practice until the outer account condition is true. The later decisions are nested because each recommendation only makes sense after the earlier condition has already chosen the correct path.
