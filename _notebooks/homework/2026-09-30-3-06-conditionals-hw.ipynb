## Homework Hack — Scam-Message Trainer Scoring Step


```python
# CODE_RUNNER: 3.6 Homework - score a message, then advise on it

def score_message(message, amount_requested, known_contact):
    risk_score = 0

    # No else is needed for these checks because a false signal simply adds zero points.
    if "gift card" in message.lower():
        risk_score = risk_score + 4

    # This is another independent signal, so it can add points without replacing another branch.
    if amount_requested > 100:
        risk_score = risk_score + 3

    # If the sender is known, this signal contributes nothing, so no else is necessary.
    if not known_contact:
        risk_score = risk_score + 2

    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 7:
        advice = "High risk: stop and verify the message through a trusted contact."
    elif risk_score >= 4:
        advice = "Elevated risk: verify the request before doing anything."
    elif risk_score >= 2:
        advice = "Some warning signs: double-check the sender and request."
    else:
        advice = "Low risk based on these three checks."

    print("Risk score:", risk_score)
    print(advice)
    return risk_score

message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False
score_message(message, amount_requested, known_contact)

print("\nExtreme test: completely clean message")
score_message("Here is the normal update", 50, True)

print("\nExtreme test: every signal")
score_message("URGENT: send gift cards now", 500, False)
```

### Reachability proof

- **High risk:** gift-card message, $500 requested, unknown sender → score 9.
- **Elevated risk:** gift-card message, $50 requested, known sender → score 4.
- **Some warning signs:** normal message, $50 requested, unknown sender → score 2.
- **Low risk:** normal message, $50 requested, known sender → score 0.

The three scoring checks do not need `else` branches because each one only contributes points when its own warning signal is present. A false check should simply leave the score unchanged.
