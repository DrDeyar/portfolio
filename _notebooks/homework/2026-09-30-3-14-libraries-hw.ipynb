## Homework Hack — Season Stats Report


```python
# CODE_RUNNER: Homework - Season Stats Report

import json
import statistics
import math
from datetime import date

# Sim-racing practice results used for this report.
games = [
    {"date": "2026-09-10", "opponent": "Apex", "points": 3, "minutes": 12},
    {"date": "2026-09-13", "opponent": "Velocity", "points": 5, "minutes": 18},
    {"date": "2026-09-17", "opponent": "Apex", "points": 4, "minutes": 15},
    {"date": "2026-09-21", "opponent": "Gridline", "points": 6, "minutes": 20},
]

def summarize_season(season):
    """
    Parameters:
        season: a list of game dictionaries containing date, opponent, points, and minutes.
    Returns:
        A dictionary with total points, average points, best game,
        and the total points scored against each opponent.
    """
    points = [game["points"] for game in season]

    points_by_opponent = {}
    for game in season:
        opponent = game["opponent"]
        points_by_opponent[opponent] = points_by_opponent.get(opponent, 0) + game["points"]

    return {
        "total_points": sum(points),
        "average_points": round(statistics.mean(points), 1),
        "best_game": max(points),
        "points_by_opponent": points_by_opponent,
        "games_played": math.ceil(len(season)),
    }

report = summarize_season(games)

next_game = date(2026, 10, 15)
report["days_until_next_game"] = (next_game - date.today()).days

print(json.dumps(report, indent=2))

requirements_txt = """pandas
Flask
# json, statistics, math, and datetime are part of Python's standard library,
# so they are not listed as packages to install.
"""

print()
print("requirements.txt:")
print(requirements_txt)
```
