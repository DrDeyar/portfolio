## Homework Hack — Crack the Final Transmission


```python
# CODE_RUNNER: Homework - Crack the Final Transmission using slicing, methods, and formatting

raw_transmission = "  2026-09-15,GHOST,THE-EAGLE-LANDS-AT-DAWN,87  "

# Clean and split the transmission into its four fields.
segments = raw_transmission.strip().split(",")
date = segments[0]
agent = segments[1]
message_code = segments[2]
confidence_str = segments[3]

print("Segments:", segments)
print("Date:", date)
print("Agent:", agent)
print("Message code:", message_code)
print("Confidence:", confidence_str)

# Use slicing to verify the year.
year_check = date[:4] == "2026"
print("Year is 2026:", year_check)

# Turn the message code into readable text and check for the word "dawn".
message = message_code.replace("-", " ")
contains_dawn = "dawn" in message.lower()
print("Readable message:", message)
print("Contains dawn:", contains_dawn)

# Format the same data in two useful output styles.
markdown_row = f"| {date} | {agent} | {message} | {confidence_str}% |"
print("Markdown row:")
print(markdown_row)

json_line = f'{{"date": "{date}", "agent": "{agent}", "message": "{message}", "confidence": {confidence_str}, "contains_dawn": {str(contains_dawn).lower()}}}'
print("JSON-style line:")
print(json_line)
```
